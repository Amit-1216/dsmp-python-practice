###`Q 1-3` Using the 'diamonds' dataset - sns.load_dataset('diamonds')
1. Create a violinplot of 'price' grouped by 'cut'.
2. Create regplot on `carat` vs `price`. and give hue on 'cut'
3. Create boxplot on 'color' and 'price'

```python
diamonds_df = sns.load_dataset('diamonds')
diamonds_df
```

```python
sns.catplot(kind='violin', data=diamonds_df, x='cut', y='price')
```

```python
## Create regplot on carat vs price. and give hue on 'cut'
## We cannot use 'hue' here, as regplot() is an axis level plot function.
## sns.regplot(data=diamonds_df, x='carat', y='price', hue='cut')

sns.lmplot(data=diamonds_df, x='carat', y='price', hue='cut')
```

```python
## Create boxplot on 'color' and 'price'
sns.catplot(kind='box', data=diamonds_df, x='color', y='price')
```

###`Q 4` Using the 'Taxis' dataset - sns.load_dataset('taxis')
4.1. Create a categorical estimate plot of the totl fare - 'total' for each payment type - 'payment'.

4.2. Create a regression plot on time of ride  vs total fare. You will need to calculate ride time using pickup and dropoff column.

4.3 Give hue on payment type. and Another plot hue on taxi 'color'. Observe the plot.

```python
taxis_df = sns.load_dataset('taxis')
taxis_df
```

```python
## 4.1. Create a categorical estimate plot of the totl fare - 'total' for each payment type - 'payment'.
sns.catplot(data=taxis_df, x="payment", y="total", kind="bar")
```

```python
## 4.2. Create a regression plot on time of ride vs total fare.
## 4.2. You will need to calculate ride time using pickup and dropoff column.

#taxis_df.info()
##taxis_df['pickup'].dt.time
taxis_df['total_duration'] = ((taxis_df['dropoff'] - taxis_df['pickup']).dt.total_seconds() / 60).round(2)
sns.lmplot(data=taxis_df, x='total_duration', y='total')
```

```python
## 4.3 Give hue on payment type. and Another plot hue on taxi 'color'.
## Observe the plot.

sns.lmplot(data=taxis_df, x='total_duration', y='total', hue='payment')
```

```python
sns.lmplot(data=taxis_df, x='total_duration', y='total', hue='color')
```

### **`Problem 5:`** Draw a strip plot and swarm plot between "gender" and "bloodpressure" w.r.t "smoker" (use as hue parameter). Also add title to the charts.

```python
insurance_df = pd.read_csv(
    'https://docs.google.com/spreadsheets/d/e/2PACX-1vQVpcVtdYdZU4zU4-lqxt-iPHkyndDWs_aqEDUu9ZodlJ48Dku0PFgdXlj2N5RCmwXJrNtZLsI_wEVf/pub?gid=220677750&single=true&output=csv',
    index_col='index'
    )
insurance_df.head()
```

```python
strip_plot = sns.catplot(data=insurance_df, x='gender', y='bloodpressure', hue='smoker', kind='strip')
strip_plot.fig.suptitle("Gender vs Blood Pressure by Smoker Status (Strip Plot)", y=1.02)
```

```python
swarm_plot = sns.catplot(data=insurance_df, x='gender', y='bloodpressure', hue='smoker', kind='swarm')
swarm_plot.fig.suptitle("Gender vs Blood Pressure by Smoker Status (Swarm Plot)", y=1.05)
```

### **`Problem 9:`** Draw a pair plot of the insurance data. Use "gender" as hue parameter.

### **`Problem 10:`** Draw a pair grid of the insurance data and use "diabetic" column as a hue parameter. Also, make the diagonal plots as box-plot, upper parts as scatter plot and the lower parts as kde plot.

```python
pg_df = insurance_df.drop(columns=['PatientID'])

pair_grid_obj = sns.PairGrid(pg_df, hue='diabetic')

pair_grid_obj.map_diag(sns.boxplot)
pair_grid_obj.map_upper(sns.scatterplot)
pair_grid_obj.map_lower(sns.kdeplot)
```

### **`Prolem 11:`** Draw a joint plot as scatter between "bloodpressure" and "bmi". Use "smoker" as hue parameter.

```python
sns.jointplot(data=insurance_df, x='bloodpressure', y='bmi', hue='smoker', kind='scatter')
```

```python
sns.jointplot(data=insurance_df, x="bloodpressure", y="bmi", kind="hex")
```

```python
joint_grid_obj = sns.JointGrid(data=insurance_df, x='bloodpressure', y='bmi')
joint_grid_obj.plot(sns.barplot,sns.histplot)
```
