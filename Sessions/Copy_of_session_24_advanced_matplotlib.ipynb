### 3D Scatter Plots

```python
fig = plt.figure(figsize=(8,8))
ax = plt.subplot(projection='3d')

ax.scatter(batters['runs'],batters['avg'],batters['strike_rate'],s=40)
ax.set_title('IPL batsman analysis')

ax.set_title('IPL batsman analysis')

ax.set_xlabel('Runs')
ax.set_ylabel('Avg')
ax.set_zlabel('SR')

# Set your labels with a smaller, safer padding
ax.set_xlabel('Runs', labelpad=10)
ax.set_ylabel('Avg', labelpad=10)
ax.set_zlabel('Strike Rate', labelpad=5)  # Keeps it inside the viewable frame

# Prevent the plot from being cut off at the edges
plt.tight_layout()

# Adding a semicolon at the final line suppresses the 'Text(...)' message
plt.show();

```

### 3D Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx, yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.log(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.tan(xx) + np.sin(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.log(xx)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

### Heatmap

```python
delivery = pd.read_csv('/content/IPL_Ball_by_Ball_2008_2022 (1).csv')
delivery.head()
```

```python
delivery = delivery[~(delivery['ballnumber'] > 6) & (delivery['batsman_run'] == 6)]
```

```python
delivery.groupby(['overs','ballnumber'])['batsman_run'].sum().unstack()
```

```python
grid = delivery.pivot_table(index='overs', columns='ballnumber', values='batsman_run',aggfunc='sum')
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.xticks(np.arange(0,6), np.arange(1,7))
plt.yticks(np.arange(0,20), np.arange(1,21))
plt.colorbar()
plt.show()
```

```python
temp_df = delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run']==6)]
```

```python
grid = temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')
```

```python
grid
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.yticks(delivery['overs'].unique(), list(range(1,21)))
plt.xticks(np.arange(0,6), list(range(1,7)))
plt.colorbar()
```
